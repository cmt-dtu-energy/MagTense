```python
mu0 = 4 * np.pi * 1e-7
tiles = Tiles(
    n=2,
    M_rem=1 / mu0,
    tile_type=[8, 8],
    color=[[1, 0, 0], [0, 0, 1]]
)

# 0: Prism 1
tiles.size = ([1.0, 0.3, 2.0], 0)
tiles.offset = ([0.0, 0.0, 0.0], 0)

# 1: Prism 2
tiles.size = ([1.0, 0.3, 2.0], 1)
tiles.offset = ([1.0, 1.0, 1.0], 1)

create_plot(tiles)
```

```python
n_points = [10, 100, 80]
seg = [0.9, 1, 0.8] / np.asarray(n_points)
pts = [[(i + 0.5) * seg[0], (j + 0.5) * seg[1], (k + 0.5) * seg[2]]
        for i in range(n_points[0])
        for j in range(n_points[1])
        for k in range(n_points[2])]
pts = np.asarray(pts, dtype=np.float64, order='F')
```

```python
updated_tiles, H_demag = run_simulation(tiles, pts)
```
