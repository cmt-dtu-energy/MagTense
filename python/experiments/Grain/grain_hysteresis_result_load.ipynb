```python
def load_hysteresis_data(filepath):
    """
    Load hysteresis data from a .npz file.
    
    Parameters:
    -----------
    filepath : str
        Path to the .npz file
    
    Returns:
    --------
    dict : Dictionary containing all loaded data
    """
    data = np.load(filepath, allow_pickle=True)
    
    result = {
        'res': data["res"],
        'M': data["res"][1][1,:,:,:],
        'BH_max': float(data["BH_max"]),
        'H_star': float(data["H_star"]),
        'B_star': float(data["B_star"]),
        'M_star': float(data["M_star"]),
        'Hc': float(data["Hc"]),
        'H_N': float(data["H_N"]),
        'runtime': float(data["runtime"]),
        'H_array': data["H_array"],
        'M_array': data["M_array"],
        'n_points': data["res"][1][1,:,:,:].shape[0]
    }
    
    return result

def load_comparison_data(base_path):
    """
    Load both CUDA and FMM results from base filename.
    
    Parameters:
    -----------
    base_path : str
        Base path without _cuda.npz or _fmm.npz suffix
    
    Returns:
    --------
    tuple : (cuda_data, fmm_data) dictionaries
    """
    cuda_data = load_hysteresis_data(base_path + "_cuda.npz")
    fmm_data = load_hysteresis_data(base_path + "_fmm.npz")
    
    print(f"CUDA: {cuda_data['n_points']} points, runtime = {cuda_data['runtime']:.2f} s, time/cell = {cuda_data['runtime']/cuda_data['n_points']:.4f} s")
    print(f"FMM:  {fmm_data['n_points']} points, runtime = {fmm_data['runtime']:.2f} s, time/cell = {fmm_data['runtime']/fmm_data['n_points']:.4f} s")
    
    return cuda_data, fmm_data

def plot_hysteresis(data, ax=None, label=None, color='k', show_markers=True):
    """
    Plot a single hysteresis curve.
    
    Parameters:
    -----------
    data : dict
        Dictionary from load_hysteresis_data
    ax : matplotlib axis, optional
        Axis to plot on. If None, creates new figure.
    label : str, optional
        Label for the curve
    color : str
        Color for the plot
    show_markers : bool
        Whether to show Hc and BH_max markers
    
    Returns:
    --------
    fig, ax : matplotlib figure and axis
    """
    if ax is None:
        fig, ax = plt.subplots(figsize=(8, 4))
    else:
        fig = ax.get_figure()
    
    # Plot main curve
    curve_label = label if label else "M/Ms"
    ax.plot(data['H_array'], data['M_array'] / Ms, ".-", color=color, 
            linewidth=1.5, markersize=6, label=curve_label)
    
    # Plot markers if requested
    if show_markers:
        ax.plot(data['Hc'], 0.0, "*", color='red', markersize=10, label="Hc")
        ax.plot(data['H_star'], data['M_star'] / Ms, "*", color='blue', 
                markersize=10, label=r"$BH_{\max}$")
    
    ax.set_xlabel("H [A/m]")
    ax.set_ylabel("M/Ms [-]")
    ax.grid(True, which="both", linestyle="--", linewidth=0.5)
    
    return fig, ax

def plot_comparison(cuda_data, fmm_data, title="Hysteresis Loop: CUDA vs FMM"):
    """
    Plot comparison of CUDA and FMM results.
    
    Parameters:
    -----------
    cuda_data : dict
        CUDA results from load_hysteresis_data
    fmm_data : dict
        FMM results from load_hysteresis_data
    title : str
        Plot title
    
    Returns:
    --------
    fig, ax : matplotlib figure and axis
    """
    fig, ax = plt.subplots(figsize=(8, 4))
    
    # Plot both curves
    ax.plot(cuda_data['H_array'], cuda_data['M_array'] / Ms, ".-r", 
            linewidth=1.5, markersize=6, 
            label=f"CUDA (t={cuda_data['runtime']:.1f}s)")
    ax.plot(fmm_data['H_array'], fmm_data['M_array'] / Ms, ".-b", 
            linewidth=1.5, markersize=6, 
            label=f"FMM (t={fmm_data['runtime']:.1f}s)")
    
    ax.set_xlabel("H [A/m]")
    ax.set_ylabel("M/Ms [-]")
    ax.set_title(title)
    ax.grid(True, which="both", linestyle="--", linewidth=0.5)
    ax.legend()
    
    fig.tight_layout()
    return fig, ax
```
