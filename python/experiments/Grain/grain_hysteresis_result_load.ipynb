```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
mu0 = 4 * np.pi * 1e-7
Ms = 1.61 / mu0
```

```python

#res_path = "../figs/Grid_rasBase_5_nGrains_5_nRef_3_dG_5e-09_fmm.npz"
#res_path = "../figs/Grid_rasBase_5_nGrains_5_nRef_3_dG_5e-09_cuda.npz"
res_path = "../figs/Grid_rasBase_5_nGrains_5_nRef_4_dG_3.75e-09_cuda.npz"

data = np.load(res_path, allow_pickle=True)

res = data["res"]       # full Micromag result object
M = res[1][1,:,:,:]
BH_max = float(data["BH_max"])
H_star = float(data["H_star"])
B_star = float(data["B_star"])
M_star = float(data["M_star"])
Hc = float(data["Hc"])
H_N = float(data["H_N"])
runtime = float(data["runtime"])
H_array = data["H_array"]
M_array = data["M_array"]

print(" number of points = ", M.shape[0])
print("runtime = ", runtime, " seconds")
```

```python
fig, ax = plt.subplots(figsize=(8, 4))

ax.plot(H_array, M_array / Ms, ".-k", linewidth=1.5, markersize=6, label="M/Ms")
ax.plot(Hc, 0.0, "r*", markersize=10, label="Hc")
ax.plot(H_star, M_star / Ms, "b*", markersize=10, label=r"$BH_{\max}$")

ax.set_xlabel("H [A/m]")
ax.set_ylabel("M/Ms [-]")
ax.set_title("Hysteresis Loop")
ax.grid(True, which="both", linestyle="--", linewidth=0.5)
ax.legend()

fig.tight_layout()
#fig.savefig(fig_path, dpi=300)
#print(f"Saved figure to: {fig_path}")
```
